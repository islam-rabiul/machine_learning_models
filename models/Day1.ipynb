```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=== 1. LOADING DATASET ===")
df = pd.read_csv('student_performance.csv')
print(df)
print("\n")
```

```python
print("=== 2. DATASET INFO ===")
print(df.info())
print("\n")
```

```python
print(df.size)
```

```python
print(df.shape)
```

```python
print("=== 3. DATA CLEANING CHECKS ===")
print("Missing values per column:\n", df.isnull().sum())
print("Number of duplicate rows:", df.duplicated().sum())
print("\n")
```

```python
print(df.head(5))
```

```python
X = df[['Hours Studied', 'Attendance (%)', 'Previous Score']]
y = df['Final Score']

print("=== 4. FEATURES (X) & TARGET (y) ===")
print("Features Matrix (X):\n", X)
print("\nTarget Vector (y):\n", y)
print("\n")
```

```python
print("=== 5. STATISTICAL SUMMARY ===")
print(df.describe())
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Hours Studied'], df['Final Score'], color='purple', s=100, label='Students')
plt.title('Hours Studied vs. Final Score')
plt.xlabel('Hours Studied')
plt.ylabel('Final Score')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

```

```python
plt.figure(figsize=(7,6))
plt.scatter(df['Hours Studied'],df['Final Score'], color='red', s=100, label='Students')
plt.title('Hours Studied vs. Final Score')
plt.xlabel('Hours Studied')
plt.ylabel('Final Score')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
