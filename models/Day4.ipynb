Create a DataFrame with annual income and spending score.
Scale the features with StandardScaler.

```python
import pandas as pd
df=pd.DataFrame({
    'annual':[1,2,3,4,5,6],
    'score':[50,60,65,48,89,91]
})
X=df[['annual','score']]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
print(X_scaled)
print(X)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
model = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster']=model.fit_predict(X_scaled)
```

```python
print(df)
```

```python
plt.scatter(df['annual'],df['score'],c=df['cluster'])
plt.xlabel('annual')
plt.ylabel('score')
plt.title('Customer_Segmentation')
plt.legend()
```

```python
import matplotlib.pyplot as plt

# Define your custom colors for each cluster ID
cluster_colors = {
    0: 'blue',  # Cluster 0 will be blue
    1: 'red',   # Cluster 1 will be red
    2: 'yellow' # Cluster 2 will be yellow
}

# Loop through each cluster
for cluster_id in df['cluster'].unique():
    cluster_data = df[df['cluster'] == cluster_id]

    plt.scatter(
        cluster_data['annual'],
        cluster_data['score'],
        c=cluster_colors[cluster_id],  # Assign the custom color here
        label=f'Cluster {cluster_id}'
    )

plt.xlabel('annual')
plt.ylabel('score')
plt.title('Customer_Segmentation')
plt.legend()
plt.show()
```
